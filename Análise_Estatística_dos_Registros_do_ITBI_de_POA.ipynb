##  2.2 Uma breve descrição do contexto e do objetivo da base:
Registros de ITBI de **2020 a 2026**, referente à compra/venda de imóveis localizados em Porto Alegre - oferece uma visão sobre o mercado imobiliário da cidade. A visualização pode apoiar a comparação de valores entre bairros, a análise da evolução dos valores ao longo do tempo (valorização) e a identificação de características que mais influenciam o valor do imóvel.

```python
# ===================================================================
# TRATAMENTO DOS DADOS
# ===================================================================

# 1. Carrega a base bruta do Drive
df_bruto = pd.read_parquet("/content/drive/MyDrive/itbi_poa_2020_2026.parquet")


# ===================================================================
# Data de Pagamento
# ===================================================================


# 2. Trata strings vazias (se houver) como NaN e remove linhas sem data_pagamento
df_limpo = df_bruto.copy()
df_limpo["data_pagamento"] = (
    df_limpo["data_pagamento"].astype(str).str.strip().replace({"": pd.NA, "None": pd.NA, "nan": pd.NA})
)
df_limpo = df_limpo.dropna(subset=["data_pagamento"])

# 3. Remove a coluna data_estimativa (errors="ignore" evita erro se você rodar a célula 2x)
df_limpo = df_limpo.drop(columns=["data_estimativa"], errors="ignore")

# 4. Converte as colunas para inteiro (Int64 suporta nulos sem quebrar)
if "cep" in df_limpo.columns:
    df_limpo["cep"] = df_limpo["cep"].astype(str).str.replace("-", "", regex=False).str.strip()

cols_int = ["n_zona_reg_imoveis", "ano_construcao", "cep"]

for col in cols_int:
    df_limpo[col] = pd.to_numeric(df_limpo[col], errors="coerce").astype("Int64")

# 5. Remove apenas os dois registros de área confirmados como inválidos.
# A combinação de _id, ano, numero_guia e área evita excluir ocorrências
# legítimas que compartilham o mesmo _id em outros anos.
registros_area_invalidos = [
    (4437, 2026, 1731796, 6796319.0),
    (52536, 2024, 1581815, 174400.0),
]
for id_registro, ano, guia, area in registros_area_invalidos:
    mascara = (
        (df_limpo["_id"] == id_registro)
        & (df_limpo["ano_referencia"] == ano)
        & (df_limpo["numero_guia"] == guia)
        & (df_limpo["area_constr_privativa"] == area)
    )
    encontrados = int(mascara.sum())
    if encontrados != 1:
        raise ValueError(
            f"Esperava exatamente um registro inválido para _id={id_registro}; "
            f"encontrei {encontrados}."
        )
    df_limpo = df_limpo.loc[~mascara].copy()

# 6. Salva dados pós-tratamento
df_limpo.to_parquet("/content/drive/MyDrive/itbi_poa_populacao_limpa.parquet", index=False)


# Verifica tipos de dados e valores faltantes
print("=== Tipos de dados ===")
print(df_limpo.dtypes)
print()
print("=== Valores faltantes (NaN diretos) ===")
print(df_limpo.isnull().sum())
```

```python
# Ordena a base limpa pelo 'numero_guia'
df_limpo_ordenado = df_limpo.sort_values(by="numero_guia")

# Identifica colunas com valores faltantes
colunas_com_nan = df_limpo_ordenado.columns[df_limpo_ordenado.isnull().any()].tolist()

print(f"Colunas com valores faltantes: {colunas_com_nan}\n")

# Mostra os 10 primeiros registros (ordenados por numero_guia) onde cada coluna é nula
for col in colunas_com_nan:
    total_nulos = df_limpo_ordenado[col].isnull().sum()
    print(f"=== Primeiros 10 registros com valor nulo na coluna: '{col}' (Total: {total_nulos}) - Ordenados por numero_guia ===")
    # Filtra e exibe os 10 primeiros registros ordenados correspondentes
    display(df_limpo_ordenado[df_limpo_ordenado[col].isnull()].head(10))
    print("\n" + "="*80 + "\n")
```

```python
# 1. Contar quantas guias (IDs únicos de guia) têm mais de um registro
guia_counts = df_limpo_ordenado["numero_guia"].value_counts()
guias_repetidas_series = guia_counts[guia_counts > 1]

qtd_guias_repetidas_unicas = len(guias_repetidas_series)
print(f"Quantidade de guias únicas que possuem mais de um registro: {qtd_guias_repetidas_unicas}\n")

# 2. Listar os 20 primeiros registros que contêm guias repetidas
# Filtra o dataframe original mantendo apenas as linhas cujas guias se repetem
guias_repetidas_lista = guias_repetidas_series.index
df_somente_repetidas = df_limpo_ordenado[df_limpo_ordenado["numero_guia"].isin(guias_repetidas_lista)]

print("=== Primeiros 20 registros que contêm guias repetidas ===")
display(df_somente_repetidas.head(20))
print("\n" + "="*80 + "\n")

# 3. Listar todas as guias que se repetem com a quantidade de registros ao lado
# Cria um DataFrame a partir da série de contagens
df_contagem_repetidas = guias_repetidas_series.reset_index()
df_contagem_repetidas.columns = ["numero_guia", "quantidade_registros"]

print("=== Todas as guias repetidas e suas respectivas quantidades ===")
display(df_contagem_repetidas)
```

*   **Quantitativa**: area_privativa --> indica a área contruída privativaconstante no documento de propriedade do imóvel em metros quadrados.

*   **Qualitativa**: finalidade_construcao --> indica a vocação construtiva da unidade imobiliária, está caracterizada pela intenção para a qual foi construída ou reformada, independentemente do seu uso atual. Ela contém categorias que cobrem desde unidades residenciais (como 'APARTAMENTO', 'APARTAMENTO DE COBERTURA', 'RESIDENCIA ISOLADA') e vagas de garagem ('ESPACO DE ESTACIONAMENTO RESIDENCIAL') até imóveis comerciais e industriais ('LOJA TERREA EM EDIFICIO', 'SALA COMERCIAL', 'DEPÓSITO / ARMAZÉM / PRÉDIO INDUSTRIAL').



# 3. Dimensionamento e seleção da amostra

Após a limpeza e a remoção dos dois registros de área confirmados como inválidos, a população analítica contém **246.340 registros**. Para dimensionar a AAS voltada à estimativa de proporções, foram adotados **95% de confiança**, **margem de erro de 5 pontos percentuais** e **p = 0,50**, valor conservador que maximiza a variância quando a proporção populacional é desconhecida. Com correção para população finita, o tamanho calculado é **384 registros**.

A amostra é selecionada sem reposição com `random_state=42`, de modo que todas as linhas elegíveis tenham a mesma probabilidade de seleção e o sorteio seja reproduzível. O tamanho de 384 atende ao cenário de proporções; a precisão para a média da área depende da variabilidade observada e é apresentada separadamente no dashboard.


```python
# 1. Extração da Amostra Aleatória Simples (AAS) sem reposição
# random_state=42 garante reprodutibilidade; sem reposição (replace=False)
df_amostra = df_limpo.sample(384, replace=False, random_state=42).copy()

# 2. Salva a amostra congelada no Drive para usar nos Painéis 2 e 3
caminho_amostra = "/content/drive/MyDrive/itbi_poa_amostra_aas.parquet"
df_amostra.to_parquet(caminho_amostra, index=False)
print(f"Amostra AAS salva em: {caminho_amostra}")
df_amostra.head()
```
